```python
########################################################
# example

import pandas as pd

# Load data
data = pd.read_csv("~/data/machine_learning/expression.csv")

# Calculate average expression
average = data["expression"].mean()
print(f"average exression: {average}")

# Identify highly expressed genes
high_expression = data[data["expression"] > average]

# Display results
print(high_expression)
```
